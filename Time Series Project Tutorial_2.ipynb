```python
import pandas as pd
```

```python
# Cargo conjunto de datos desde Kaggle
data = pd.read_csv('River_Arno.csv')
data.head()
```

```python
data.info()
```

```python
target = ['Hydrometry_Nave_di_Rosano']
features = [feature for feature in data.columns if feature not in target]
data.head()
```

```python
data = data.set_index('Date')
data.head()
```

```python
data = data[target]
data.head()
```

```python
# Relleno valores nulos de variable target
data = data.fillna(method='ffill')
data.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axis = plt.subplots(figsize=(10, 5))
sns.lineplot(data=data)

plt.tight_layout()
plt.show()
```

```python
from statsmodels.tsa.seasonal import seasonal_decompose

decomposition = seasonal_decompose(data, period = 12)
decomposition
```

```python
trend = decomposition.trend

fig, axis = plt.subplots(figsize = (6, 4))

sns.lineplot(data = data)
sns.lineplot(data = trend)

plt.tight_layout()

plt.show()
```

```python
from statsmodels.tsa.stattools import adfuller

def test_stationarity(timeseries):
    print("Dickey-Fuller test results:")
    dftest = adfuller(timeseries, autolag = "AIC")
    dfoutput = pd.Series(dftest[0:4], index = ["Test Statistic", "p-value", "#Lags Used", "Number of Observations Used"])
    for key,value in dftest[4].items():
        dfoutput["Critical Value (%s)"%key] = value
    return dfoutput

test_stationarity(data)
```

```python
from statsmodels.graphics.tsaplots import plot_acf
plt.figure(figsize=(6, 4))
plot_acf(data)
plt.tight_layout()
plt.show()
```

```python
ts_stationary = data.diff().dropna()

test_stationarity(ts_stationary)
```

```python
train = ts_stationary[:int(len(data)*0.8)]
test = ts_stationary[int(len(data)*0.8):]

print (train.shape)
print (test.shape)
```

```python
import numpy as np
from pmdarima import auto_arima

model = auto_arima(train, seasonal = True, trace = True, m = 12)
```

```python
model.summary()
```

```python
# Número de pasos a predecir = largo del conjunto de prueba
n_periods = len(test)

# Generar predicción
forecast = model.predict(n_periods=n_periods)

# Convertimos la predicción a una Serie con el mismo índice que test
forecast = pd.Series(forecast, index=test.index)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.plot(train, label='Train')
plt.plot(test, label='Test', color='blue')
plt.plot(forecast, label='Forecast', color='yellow')
plt.legend()
plt.title("ARIMA Forecast vs Real Test Data")
plt.show()
```
